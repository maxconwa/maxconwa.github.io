## The update

Steepest descent takes the step $x_{k+1} = x_k - \alpha \nabla f(x_k)$. The whole method is one line; everything interesting is in what $\alpha$ and the curvature of $f$ do to it.

Edit any cell below and run it.

```python
import numpy as np

def f(x):
    """A bowl with an awkward aspect ratio."""
    return 0.5 * x[0] ** 2 + 5.0 * x[1] ** 2

def grad(x):
    return np.array([x[0], 10.0 * x[1]])

x = np.array([9.0, 1.0])
path = [x.copy()]
for step in range(40):
    x = x - 0.08 * grad(x)
    path.append(x.copy())

path = np.array(path)
print(f"start f = {f(path[0]):.3f}")
print(f"end   f = {f(path[-1]):.6f}")
```

## Why it zig-zags

The Hessian here has eigenvalues 1 and 10. A step size that is stable for the steep direction is far too small for the shallow one.

```python
import matplotlib.pyplot as plt

grid = np.linspace(-10, 10, 200)
gx, gy = np.meshgrid(grid, np.linspace(-2, 2, 200))

fig, ax = plt.subplots(figsize=(6, 2.6))
ax.contour(gx, gy, 0.5 * gx ** 2 + 5.0 * gy ** 2, levels=18, linewidths=0.6)
ax.plot(path[:, 0], path[:, 1], marker="o", markersize=2.5, linewidth=1.2)
ax.set_xlabel("$x_0$")
ax.set_ylabel("$x_1$")
ax.set_title("Descent zig-zags across the narrow axis")
fig.tight_layout()
plt.show()
```
