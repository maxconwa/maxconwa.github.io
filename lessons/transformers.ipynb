## Attention

$$\mathrm{Attention}(Q, K, V) = \mathrm{softmax}\!\left(\frac{QK^\top}{\sqrt{d_k}}\right) V$$

> This lesson needs PyTorch, which cannot run in the browser. The outputs below were saved from a real run; open it in Colab to execute it yourself.

```python
import torch
import torch.nn as nn

block = nn.TransformerEncoderLayer(d_model=64, nhead=4, batch_first=True)
x = torch.randn(2, 16, 64)
print(block(x).shape)
print(sum(p.numel() for p in block.parameters()), "parameters")
```
