## The chain rule, arranged

For a layer $h = \tanh(Wx + b)$ the backward pass is

$$\frac{\partial L}{\partial W} = x \left(\frac{\partial L}{\partial h} \odot (1 - h^2)\right)^\top$$

Every framework you will use is this, bookkept.

```python
import numpy as np

rng = np.random.default_rng(0)
X = rng.normal(size=(64, 3))
y = (X @ np.array([2.0, -1.0, 0.5]) > 0).astype(float)[:, None]

W1, b1 = rng.normal(size=(3, 8)) * 0.5, np.zeros(8)
W2, b2 = rng.normal(size=(8, 1)) * 0.5, np.zeros(1)

for step in range(400):
    h = np.tanh(X @ W1 + b1)
    p = 1 / (1 + np.exp(-(h @ W2 + b2)))
    loss = -np.mean(y * np.log(p + 1e-9) + (1 - y) * np.log(1 - p + 1e-9))

    d_out = (p - y) / len(X)
    dW2, db2 = h.T @ d_out, d_out.sum(0)
    d_h = (d_out @ W2.T) * (1 - h ** 2)
    dW1, db1 = X.T @ d_h, d_h.sum(0)

    for param, grad in ((W1, dW1), (b1, db1), (W2, dW2), (b2, db2)):
        param -= 0.5 * grad

    if step % 100 == 0:
        print(f"step {step:3d}  loss {loss:.4f}")

print(f"final accuracy {((p > 0.5) == y).mean():.3f}")
```
