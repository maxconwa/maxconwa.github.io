## The problem

A POMDP is a tuple $(S, A, O, T, Z, R, \gamma)$. The robot never sees $s$; it maintains a belief $b(s)$ and plans over beliefs.

> This lesson drives MuJoCo and a ROS 2 stack, so it does not run in the browser. Outputs below are from a run on the lab's H1.

```python
import mujoco
from lab.planning import BeliefTreeSearch

model = mujoco.MjModel.from_xml_path("assets/h1_table.xml")
planner = BeliefTreeSearch(model, horizon=12, particles=256)
result = planner.solve(goal="grasp_mug")

print(f"expected return {result.value:.2f}")
print(f"nodes expanded  {result.expanded}")
```
